# Minor Project: Flask Mini Project
### Student Registration System
This notebook builds a simple web application using Python Flask. It satisfies all project requirements:
- Creates a Flask application
- Implements 3 working routes (`/`, `/register`, `/about`)
- Employs GET and POST routes
- Performs basic input validation with friendly error messages
- Renders a meaningful result using simple HTML structures

```python
from flask import Flask, render_template_string, request
from threading import Thread
from google.colab import output

app = Flask(__name__)

# In-memory array storage acting as our simple runtime session database
registered_students = []

# Simple inline HTML UI Templates
INDEX_HTML = """
<!DOCTYPE html>
<html>
<head>
    <title>Student Registration</title>
</head>
<body style=\"font-family: sans-serif; max-width: 450px; margin: 30px auto; padding: 20px; border: 1px solid #ccc; border-radius: 5px;\">
    <h2>Student Registration Form</h2>
    <p><a href=\"/about\">View System Info / About App</a></p>

    {% if error %}
        <div style=\"color: red; background: #fee; padding: 10px; margin-bottom: 15px; border: 1px solid red;\">
            <strong>Error:</strong> {{ error }}
        </div>
    {% endif %}

    <form action=\"/register\" method=\"POST\">
        <div style=\"margin-bottom: 10px;\">
            <label style=\"display:block;\">Full Name:</label>
            <input type=\"text\" name=\"name\" style=\"width: 100%; padding: 5px;\">
        </div>
        <div style=\"margin-bottom: 10px;\">
            <label style=\"display:block;\">Email Address:</label>
            <input type=\"text\" name=\"email\" style=\"width: 100%; padding: 5px;\">
        </div>
        <div style=\"margin-bottom: 15px;\">
            <label style=\"display:block;\">Select Course:</label>
            <select name=\"course\" style=\"width: 100%; padding: 5px;\">
                <option value=\"\">-- Choose Course --</option>
                <option value=\"Python Basics\">Python Basics</option>
                <option value=\"Flask Web Dev\">Flask Web Dev</option>
            </select>
        </div>
        <button type=\"submit\" style=\"background: blue; color: white; padding: 8px; width: 100%; border: none; cursor: pointer;\">
            Submit Registration
        </button>
    </form>
</body>
</html>
"""

RESULT_HTML = """
<!DOCTYPE html>
<html>
<head>
    <title>Success</title>
</head>
<body style=\"font-family: sans-serif; max-width: 450px; margin: 30px auto; padding: 20px; border: 1px solid green; background: #f9fff9; border-radius: 5px;\">
    <h2 style=\"color: green;\">ð Success!</h2>
    <p>Student registration data has been captured.</p>
    <div style=\"background: white; padding: 10px; border: 1px solid #ddd;\">
        <p><strong>Name:</strong> {{ student.name }}</p>
        <p><strong>Email:</strong> {{ student.email }}</p>
        <p><strong>Course:</strong> {{ student.course }}</p>
    </div>
    <br>
    <a href=\"/\">← Register Another Student</a> | <a href=\"/about\">View System Status</a>
</body>
</html>
"""

# Route 1 (GET): Homepage Form Display
@app.route('/')
def home():
    return render_template_string(INDEX_HTML)

# Route 2 (POST): Form Handler & Input Checker
@app.route('/register', methods=['POST'])
def register():
    name = request.form.get('name', '').strip()
    email = request.form.get('email', '').strip()
    course = request.form.get('course', '').strip()

    # Basic Validation Rule Check
    if not name or not email or not course:
        return render_template_string(INDEX_HTML, error="All form fields are mandatory!")

    if "@" not in email:
        return render_template_string(INDEX_HTML, error="Please enter a valid email structure.")

    student_record = {"name": name, "email": email, "course": course}
    registered_students.append(student_record)

    return render_template_string(RESULT_HTML, student=student_record)

# Route 3 (GET): Dynamic System Info Page
@app.route('/about')
def about():
    count = len(registered_students)
    return f"""
    <body style=\"font-family: sans-serif; padding: 30px;\">
        <h2>About Registration App</h2>
        <p>This beginner project tracks live form submissions dynamically.</p>
        <p><strong>Live Database Metric:</strong> {count} students registered in this session.</p>
        <a href=\"/\">← Go Back Form Screen</a>
    </body>
    """

def run_flask():
    app.run(host="0.0.0.0", port=5001,
            debug=False, use_reloader=False)

Thread(target=run_flask, daemon=True).start()

output.serve_kernel_port_as_window(5001)
```

### GET route

```python
from IPython.display import display, HTML

def test_my_flask_project(action_route, method="GET", form_data=None):
    """Simulates internal browser requests to bypass machine security walls."""
    with app.test_client() as client:
        if method == "POST":
            response = client.post(action_route, data=form_data)
        else:
            response = client.get(action_route)

        # Instantly render the raw HTML output right inside your notebook!
        display(HTML(response.data.decode('utf-8')))
```

```python
# View the initial clean structural form UI
test_my_flask_project(action_route='/', method='GET')
```
